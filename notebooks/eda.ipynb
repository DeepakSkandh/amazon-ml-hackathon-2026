```python
# ============================================================
# EDA REPORT — CORE STATISTICS
# ============================================================

def source_summary(df, name):
    return {
        "source": name,
        "rows": len(df),
        "unique_ids": df["entity_id"].nunique(),
        "unique_names": df["business_name"].nunique(),
        "unique_addresses": df["business_address"].nunique(),
        "duplicate_ids": df["entity_id"].duplicated().sum(),
        "name_missing": df["business_name"].isna().sum(),
        "address_missing": df["business_address"].isna().sum(),
        "country_missing": df["country"].isna().sum(),
        "countries": df["country"].value_counts().to_dict(),
        "name_median_length": df["business_name"].fillna("").str.len().median(),
        "address_median_length": df["business_address"].fillna("").str.len().median(),
    }


summaries = [
    source_summary(train_s1, "S1"),
    source_summary(train_s2, "S2"),
    source_summary(train_s3, "S3"),
]

pd.DataFrame(summaries)
```
